```python
import geopandas as gpd
import os
from pyogrio import read_info
```

```python
path = r'../data\input_data\Vectore_file\Assam\vb_soi_as.shp'

# Inspect your datasets
# Look at the metadata first, without loading everything.
info = read_info(path)
print(info["crs"], info["geometry_type"], info["features"])
print(info["fields"])
```

```python
gdf = gpd.read_file(path)
gdf.head(5)
```

```python
gdf.crs
```

```python
gdf["area_sqkm"] = gdf.area / 1000000
```

```python
gdf = gdf.drop(columns=['id', 'objectid', 'gram_panch',
       'gram_pan_1', 'block', 'bkcode', 'subdistric', 'sdcode', 'sub_distri','dtcode', 'district_h', 'state', 'stcode', 'uqcode',
       'country', 'total_hous', 'total_popu', 'avg_househ',
       'total_geog', 'total_male', 'total_fema', 'tapwater_t', 'tapwater_u',
       'covered_we', 'covered__1', 'covered__2', 'uncovered_', 'uncovere_1',
       'uncovere_2', 'handpump_s', 'hand_pump_', 'hand_pum_1', 'tubewell_b',
       'tube_wells', 'tube_wel_1', 'spring_sta', 'spring_fun', 'spring_f_1',
       'river_cana', 'tank_pond_', 'others_sta', 'closed_dra', 'open_drain',
       'village_pi', 'forest_are', 'area_under', 'barren_unc', 'permanent_',
       'land_under', 'culturable', 'fallows_la', 'current_fa', 'net_area_s',
       'total_unir', 'area_irrig', 'canals_are', 'wells_tube', 'tanks_lake',
       'waterfall_', 'other_sour', 'nearest_to', 'nearest__1', 'shape_leng',
       'shape_le_1', 'ds_name', 'src_agency', ])
```
