```python
import geopandas as gpd
import os
from pyogrio import read_info
```

```python
path = r'../data\input_data\Vectore_file\Assam\vb_soi_as.shp'

# Inspect your datasets
# Look at the metadata first, without loading everything.
info = read_info(path)
print(info["crs"], info["geometry_type"], info["features"])
print(info["fields"])
```

```python
gdf = gpd.read_file(path)
gdf.head(5)
```

```python
print(gdf['total_urba'].unique())
print(gdf['total_urba'].value_counts())
```

### Drop the urban features from analysis

```python
gdf_rural = gdf[gdf['total_urba']=="Rural"]
gdf_rural.shape
```

```python
# in filtered gdf with manual inspection of village 
# name with Reserved Forest or R F shows mixed output. 
# Some of them has good amout of developed builtup area. 
# so its better to filtout this latter not on by the name or forest availablity.

# save the filtered village boundary file
gdf_rural.to_crs(epsg=4326, inplace=True)
```

```python
out_dir = "../data/output_data/vectore"
os.makedirs(out_dir, exist_ok=True)
out_file = os.path.join(out_dir, 'filtered_village.shp')
gdf_rural.to_file(out_file, driver='ESRI Shapefile')
```
